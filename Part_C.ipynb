Steps performed:

Tokenize the text.

Remove stopwords using spaCy’s built-in stopword list.

Lemmatize remaining tokens.

Keep only tokens whose POS tag is NOUN or VERB.

Print the final list of lemmas.

```python
## part_c

# CS5710 - Machine Learning
#Home Assignment 4 - Part C (Coding)

#This script contains solutions for:
    #Q1. Tokenization, stopword removal, lemmatization, and POS filtering.
    
# Library: spaCy (en_core_web_sm)
###
import spacy

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")

text = "John enjoys playing football while Mary loves reading books in the library."

# Process text with spaCy (tokenization + POS + lemmas all at once)
doc = nlp(text)

# Built-in spaCy stopword list
stop_words = nlp.Defaults.stop_words

# Keep only tokens that:
#  - are alphabetic
#  - are not stopwords
#  - and are nouns or verbs
filtered_lemmas = [
    token.lemma_
    for token in doc
    if token.is_alpha
    and token.text.lower() not in stop_words
    and token.pos_ in ("NOUN", "VERB")
]

print("Final Output Tokens:", filtered_lemmas)

```

Interpretation:

Verbs like “enjoys”, “loves”, “reading” are reduced to their base forms: enjoy, love, read.

Plural nouns like “books” become book.

Only content words (nouns & verbs) are kept; function words like “while”, “in”, “the” are removed as stopwords.

```python
# PART C – Q2
# ---------------------------------
# Named Entity Recognition (NER)
# + pronoun ambiguity warning
# ---------------------------------

import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

text = "Chris met Alex at Apple headquarters in California. He told him about the new iPhone launch."

doc = nlp(text)

# 1) Named Entity Recognition
print("Named Entities:")
for ent in doc.ents:
    print(f"{ent.text}  ->  {ent.label_}")

# 2) Pronoun ambiguity warning
pronouns = {"he", "she", "they", "him", "her", "them"}

if any(token.text.lower() in pronouns for token in doc):
    print("\nWarning: Possible pronoun ambiguity detected!")
```

Interpretation:

The model correctly identifies people, organizations, locations, and products.

The warning highlights that pronouns like “He” and “him” could refer to either Chris or Alex, which is a common source of ambiguity for NLP systems.
